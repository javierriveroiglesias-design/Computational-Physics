### Introduction to Vectors and NumPy

Let's first see how to create vectors in Python using the NumPy library. We store the vector components in a NumPy array.  Run the following code cell to execute it by pressing Shift-Enter or clicking the play button.

```python
import numpy as np
import plotly.graph_objects as go

# Define 3D vectors using NumPy arrays
A = np.array([1, 2, 3])
B = np.array([-1, 0, 2])

# Display the vectors
print("Vector A = ", A)
print("Vector B = ", B)
```

## Magnitude of a Vector

The magnitude of a vector, also known as its Euclidean norm or "length," is calculated using the Pythagorean theorem. For a 3D vector $ \mathbf{A} = [A_x, A_y, A_z] $, its magnitude is given by:

$$ A=|\mathbf{A}| = \sqrt{A_x^2 + A_y^2 + A_z^2} .$$

In NumPy, you can use NumPy's Linear Algebra module and the `norm()` function to calculate a vector's magnitude. The command looks like this:  `np.linalg.norm()` .

```python
import numpy as np

# Calculate the magnitude of A and B
A_mag = np.linalg.norm(A)
B_mag = np.linalg.norm(B)

# Display the results
print(f"Magnitude of A = {A_mag:.2f}")
print(f"Magnitude of B = {B_mag:.2f}")
```

## Dot Product

The dot product (also known as the scalar product) takes two vectors and returns a single scalar value. It's often used to find the angle between two vectors or to project one vector onto another.

For two vectors $ \mathbf{A} = [A_1, A_2, A_n] $ and $ \mathbf{B} = [B_1, B_2, .B_3] $, the dot product is defined as:

$$ \mathbf{A} \cdot \mathbf{B} = \sum_{i=1}^{n} A_i B_i = A_1 B_1 + A_2 B_2 +A_3 B_3$$

In NumPy, you can use `np.dot(A,B)` or the ` @` operator to write `A @ B`:

```python
import numpy as np

# Finding the dot product between A and B
prod = np.dot(A,B)
print(f"Dot product A and B = {prod}")

# This example shows how to use the @ operator for dot products
prod2 = A @ B
print(f"We get the same result using '@':  A @ B = {prod2}")
```

## Angle Between Two Vectors

The dot product can be written in two ways:
$$ \mathbf{A} \cdot \mathbf{B} = \sum_{i=1}^{n} A_i B_i = A_1 B_1 + A_2 B_2 +A_3 B_3$$
and
$$ \mathbf{A} \cdot \mathbf{B}  = A B \cos \theta. $$

The Python command `np.dot()` uses the first method to calculate the dot product. We can then use the second definiton to solve for the angle $\theta$ once the dot product is known:
$$\theta = \cos^{-1} \left( \frac{ \mathbf{A} \cdot \mathbf{B}}{AB} \right) $$


```python
import numpy as np

# Calculate the magnitude of v1 and v2
A_mag = np.linalg.norm(A)
B_mag = np.linalg.norm(B)

# Calculate the angle of v1 and v2
theta = np.acos(np.dot(A,B) / (A_mag * B_mag))

# Convert to degrees
theta_deg = np.rad2deg(theta)

# Display the results
print(f"Angle between v1 and v2 = {theta_deg:.2f} degrees")
```

## Create a Unit Vector $\widehat{\mathbf{a}}$ from a Vector $\mathbf{A}$

To create a unit vector pointing in the same direction as a vector $\mathbf{A}$, we simply divide the vector by its magnitude to force it to have unit length:
$$ \widehat{\mathbf{a}}  = \frac{\mathbf{A}}{A},$$
where we use the "hat" ($\;\widehat{}\;$) to designate a unit vector.



```python
# create a unit vector that points in the direction of vector A
a = A / A_mag
print(f"If A = {A}, the corresponding unit vector is {a}")
```

### Cross Product

The cross product (also known as the vector product) takes two 3D vectors and returns a new 3D vector that is perpendicular to both original vectors. The magnitude of the cross product is related to the area of the parallelogram formed by the two vectors.

For two 3D vectors $ \mathbf{A} = [A_x, A_y, A_z] $ and $ \mathbf{B} = [B_x, B_y, B_z] $, the cross product is defined as:

$ \mathbf{A} \times \mathbf{B} = [ (A_y B_z - A_z B_y), -(A_x B_z - A_z B_x ), (A_x B_y - A_y B_x) ] $

In NumPy, you can use `np.cross(A,B)`.

```python
import numpy as np

# Calculate the cross product C = A x B
C = np.cross(A,B)

print(f"Cross product A x B = {C}")
```

```python
# Import required libraries
import numpy as np
import plotly.graph_objects as go

# Function to add a vector to the plot (FIXED - removed row/col parameters)
def add_vector(fig, vector, color, name):
    # Add arrow (cone)
    fig.add_trace(go.Cone(
        x=[vector[0]],
        y=[vector[1]],
        z=[vector[2]],
        u=[vector[0]],
        v=[vector[1]],
        w=[vector[2]],
        colorscale=[[0, color], [1, color]],
        showscale=False,
        sizemode="absolute",
        sizeref=0.5,
        name=f'{name}: ({vector[0]:.1f}, {vector[1]:.1f}, {vector[2]:.1f})'
    ))

    # Add the vector shaft as a line
    fig.add_trace(go.Scatter3d(
        x=[0, vector[0]],
        y=[0, vector[1]],
        z=[0, vector[2]],
        mode='lines',
        line=dict(color=color, width=10),
        showlegend=False
    ))

    # Add label at the tip of the vector
    fig.add_trace(go.Scatter3d(
        x=[vector[0]],
        y=[vector[1]],
        z=[vector[2]],
        mode='text',
        text=[name],
        textposition='top center',
        textfont=dict(size=16, color=color, family='Arial Black'),
        showlegend=False
    ))

# Function to create a single plot
def create_cross_product_plot(A, B, title):
    # Calculate the cross product C = A × B
    C = np.cross(A, B)

    # Create the 3D plot
    fig = go.Figure()

    # Add all three vectors to the plot
    add_vector(fig, A, 'red', 'A')
    add_vector(fig, B, 'blue', 'B')
    add_vector(fig, C, 'green', 'C=A×B')

    # Add origin point
    fig.add_trace(go.Scatter3d(
        x=[0],
        y=[0],
        z=[0],
        mode='markers',
        marker=dict(size=10, color='black'),
        name='Origin',
        showlegend=False
    ))

    # Set axis limits to make the plot symmetric
    all_vectors = np.array([A, B, C])
    max_val = max(np.max(np.abs(all_vectors)) * 1.3, 1)

    # Update layout
    fig.update_layout(
        title=f'{title}<br>A={A}, B={B}, C={C}',
        scene=dict(
            xaxis=dict(title='X axis', range=[-max_val, max_val]),
            yaxis=dict(title='Y axis', range=[-max_val, max_val]),
            zaxis=dict(title='Z axis', range=[-max_val, max_val]),
            aspectmode='cube'
        ),
        showlegend=True,
        width=900,
        height=700
    )

    return fig

# ============================================================================
# Test Case 1: Original vectors from your code
# ============================================================================
print("="*70)
print("TEST CASE 1: Original Vectors")
print("="*70)
A1 = np.array([2, 1, 0])
B1 = np.array([1, 2, 1])
C1 = np.cross(A1, B1)
print(f"A = {A1}")
print(f"B = {B1}")
print(f"C = A × B = {C1}")
print(f"|C| = {np.linalg.norm(C1):.4f}")
print()

fig1 = create_cross_product_plot(A1, B1, "Case 1: General Vectors")
fig1.show()

# ============================================================================
# Test Case 2: Perpendicular unit vectors (Right-hand rule example)
# ============================================================================
print("="*70)
print("TEST CASE 2: Perpendicular Unit Vectors (x̂ and ŷ)")
print("="*70)
A2 = np.array([1, 0, 0])
B2 = np.array([0, 1, 0])
C2 = np.cross(A2, B2)
print(f"A = {A2} (unit vector along x)")
print(f"B = {B2} (unit vector along y)")
print(f"C = A × B = {C2} (should be unit vector along z)")
print(f"|C| = {np.linalg.norm(C2):.4f}")
print("✓ This demonstrates the right-hand rule!")
print()

fig2 = create_cross_product_plot(A2, B2, "Case 2: Perpendicular Unit Vectors (x̂ × ŷ = ẑ)")
fig2.show()

# ============================================================================
# Test Case 3: Parallel vectors (cross product = 0)
# ============================================================================
print("="*70)
print("TEST CASE 3: Parallel Vectors")
print("="*70)
A3 = np.array([2, 4, 6])
B3 = np.array([1, 2, 3])
C3 = np.cross(A3, B3)
print(f"A = {A3}")
print(f"B = {B3} (B = 0.5 * A, so they're parallel)")
print(f"C = A × B = {C3} (should be zero vector)")
print(f"|C| = {np.linalg.norm(C3):.4f}")
print("✓ Parallel vectors have zero cross product!")
print()

fig3 = create_cross_product_plot(A3, B3, "Case 3: Parallel Vectors (C = 0)")
fig3.show()

# ============================================================================
# Test Case 4: Anti-commutativity (B × A = -A × B)
# ============================================================================
print("="*70)
print("TEST CASE 4: Anti-commutativity")
print("="*70)
A4 = np.array([3, -2, 1])
B4 = np.array([1, 1, -2])
C4_AB = np.cross(A4, B4)
C4_BA = np.cross(B4, A4)
print(f"A = {A4}")
print(f"B = {B4}")
print(f"A × B = {C4_AB}")
print(f"B × A = {C4_BA}")
print(f"-(B × A) = {-C4_BA}")
print(f"A × B = -(B × A)? {np.allclose(C4_AB, -C4_BA)}")
print()

# Show both A × B and B × A
fig4 = go.Figure()
add_vector(fig4, A4, 'red', 'A')
add_vector(fig4, B4, 'blue', 'B')
add_vector(fig4, C4_AB, 'green', 'A×B')
add_vector(fig4, C4_BA, 'purple', 'B×A')
fig4.add_trace(go.Scatter3d(x=[0], y=[0], z=[0], mode='markers',
                            marker=dict(size=10, color='black'), showlegend=False))

all_vectors = np.array([A4, B4, C4_AB, C4_BA])
max_val = max(np.max(np.abs(all_vectors)) * 1.3, 1)
fig4.update_layout(
    title=f'Case 4: Anti-commutativity<br>A×B={C4_AB}, B×A={C4_BA} (opposite directions!)',
    scene=dict(
        xaxis=dict(title='X axis', range=[-max_val, max_val]),
        yaxis=dict(title='Y axis', range=[-max_val, max_val]),
        zaxis=dict(title='Z axis', range=[-max_val, max_val]),
        aspectmode='cube'
    ),
    showlegend=True,
    width=900,
    height=700
)
fig4.show()

# ============================================================================
# Test Case 5: Your turn! Modify these vectors
# ============================================================================
print("="*70)
print("TEST CASE 5: Custom Vectors (MODIFY THESE!)")
print("="*70)
A5 = np.array([1, 2, 3])  # <-- Change these values!
B5 = np.array([4, 5, 6])  # <-- Change these values!
C5 = np.cross(A5, B5)
print(f"A = {A5}")
print(f"B = {B5}")
print(f"C = A × B = {C5}")
print(f"|C| = {np.linalg.norm(C5):.4f}")
print()

# Check if perpendicular
A_dot_C = np.dot(A5, C5)
B_dot_C = np.dot(B5, C5)
print("Perpendicularity check:")
print(f"A · C = {A_dot_C:.6f} (should be ≈ 0)")
print(f"B · C = {B_dot_C:.6f} (should be ≈ 0)")
print()

fig5 = create_cross_product_plot(A5, B5, "Case 5: Your Custom Vectors")
fig5.show()

# ============================================================================
# Summary and Key Insights
# ============================================================================
print("="*70)
print("KEY INSIGHTS FROM CROSS PRODUCT:")
print("="*70)
print("1. ✓ C = A × B is PERPENDICULAR to both A and B")
print("2. ✓ Right-hand rule: point fingers along A, curl toward B, thumb points to C")
print("3. ✓ Anti-commutative: A × B = -(B × A)")
print("4. ✓ Parallel vectors: A × A = 0 (or A × (kA) = 0)")
print("5. ✓ Magnitude: |A × B| = |A||B|sin(θ)")
print("6. ✓ C points in direction determined by right-hand rule")
print("="*70)
```
